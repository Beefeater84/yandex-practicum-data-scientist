```python
from sklearn.datasets import make_regression
```

```python
# Генерация данных:
# 200 примеров (наблюдений)
# 5 признаков
# шум с дисперсией 15
# фиксированное случайное зерно для воспроизводимости

X, y = make_regression(n_samples=200, 
                       n_features=5,
                       noise=15,
                       random_state=42)

# Вывод размеров массивов
print("Размеры X:", X.shape)
```

## Визуализация датасета


```python
import matplotlib.pyplot as plt
```

Чтобы увидеть, как целевая переменная `y` связана с первым признаком X[0], построим диаграмму рассеяния (scatter plot). На ней будет видно, как изменится y, когда меняется значение первого признака.


```python
# Построение диаграммы рассеяния
plt.figure(figsize=(8, 6))
# На оси X - значения первого признака, на оси Y - целевая переменная
plt.scatter(X[:, 0], y)
plt.title('Зависимость целевой переменной от признака 0')
plt.xlabel('Признак 0')
plt.ylabel('Целевая переменная (y)')
# Корректируем отступы чтобы чтобы все было видно
plt.tight_layout()
plt.show()
```

Для более полного анализа нужно посмотреть, как целевая переменная связана со всеми признаками. Так можно сравнить их влияние и определить, какие признаки лучше всего объясняют вариацию целевой переменной.


```python
# Создаём 5 графиков рассеяния для первых 5 признаков

fig, axes = plt.subplots(1, 5, figsize=(20, 4))

for i in range(5):
    axes[i].scatter(X[:, i], y)
    axes[i].set_title(f'Feature {i}')
    axes[i].set_xlabel(f'X')
    axes[i].set_ylabel('Target (y)')

plt.tight_layout()
plt.show()
```

## Потроение модели линейной регрессии


### Инициализация модели


```python
from sklearn.linear_model import LinearRegression
```

LinearRegression решает Линейную регрессию аналитически - с помощью системы управнений. Но зато за одну итерацию.


```python
# Инициируем модель линейной регрессии
model = LinearRegression()
```

### Обучение модели


Подробнее о работе метода:

- Метод .fit() запускает процесс обучения модели.
- Внутри происходит вычисление коэффициентов (весов), которые минимизируют среднеквадратичную ошибку (MSE) между реальными значениями y и предсказанными значениями модели.
- Метод аналитически решает систему уравнений, используя формулу наименьших квадратов (как вы уже изучали в NumPy).
- Обучение происходит за один шаг — без итераций, что очень быстро и эффективно для небольших и средних наборов данных.


```python
model.fit(X, y)
```
